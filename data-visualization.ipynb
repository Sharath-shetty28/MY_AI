```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Student": ["Sharath", "Sneha", "Anil", "Rahul", "Divya"],
    "Math": [85, 90, 78, 88, 92],
    "Science": [80, 85, 75, 89, 94]
})

# 1. Bar chart of Math scores
plt.bar(df["Student"], df["Math"])
plt.title("Math Scores")
plt.show()

# 2. Line plot of Science scores
plt.plot(df["Student"], df["Science"])
plt.title("Science Scores")
plt.show()

# 3. Heatmap for subject correlation
sns.heatmap(df[["Math", "Science"]].corr(), annot=True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 8, 7]

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```

```python
plt.subplot(2,2,1)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.subplot(2,2,2)
plt.plot(y,x)
```

**figure controlling of canvas**

```python
fig=plt.figure()
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2=fig.add_axes([0.5,0.4,0.4,0.6])
axis2.plot(x,y)
#0.1 is 10%
```

```python
fig=plt.figure(figsize=(9,5),dpi=520)
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
#plt.saveifg("basic.png")
axis2=fig.add_axes([0.5,0.4,0.4,0.6])
axis2.plot(x,y)
#figsize as width X height 
#dpi for clear vision
```

```python
labels = ['Math', 'Science', 'English']
sizes = [30, 40, 30]

plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=140)
plt.title("Subject Distribution")
plt.axis('equal')  # Circle shape
plt.show()
```

```python
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    "Marks": [85, 92, 78, 88, 91],
    "Study Hours": [5, 6, 4, 5.5, 6.5]
})

sns.scatterplot(x="Study Hours", y="Marks", data=df)
plt.title("Marks vs Study Hours")
plt.show()
```

```python
data = {
    "Math": [85, 90, 78],
    "Science": [88, 92, 76],
    "English": [80, 85, 79]
}

df = pd.DataFrame(data)

sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Subject Correlation Heatmap")
plt.show()
```
