**dataframes**

```python
import pandas as pd

data = {
    "Name": ["Sharath", "Rahul", "Sneha"],
    "Age": [21, 22, 23],
    "City": ["Udupi", "Bangalore", "Mysore"]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())        # Show first 5 rows
print(df.tail())        # Show last 5 rows
print(df.columns)       # Show column names
print(df["Name"])       # Show 'Name' column
print(df["Age"].mean()) # Average age
print(df[["Name","City"]])
```

```python
import pandas as pd

data = {
    "Student": ["Anil", "Divya", "Kiran", "Sneha"],
    "Marks": [85, 92, 76, 88],
    "Subject": ["Math", "Science", "History", "Math"]
}

df = pd.DataFrame(data)

# 1. Print top 2 rows
print(df.head(2))

# 2. Print average marks
print("Average:", df["Marks"].mean())

# 3. Show students who scored > 80
print(df[df["Marks"] > 80])
```
