# 1) Create a table with the 5-number summary of all the numeric attributes. 

```python
df.replace('?', np.nan, inplace=True)
df = df.apply(pd.to_numeric, errors='coerce')  


numeric_cols = df.select_dtypes(include=np.number)


summary_df = numeric_cols.describe(percentiles=[0.25, 0.5, 0.75]) \
                        .loc[['min', '25%', '50%', '75%', 'max']] \
                        .round(2)

summary_df.index = ['Min', 'Q1', 'Median', 'Q3', 'Max']

summary_df = summary_df.T.dropna(how='all')


print(summary_df)
```

# 2) For each of the numeric attributes (proteins up to vitamins), identify and replace all 
missing data (indicated with -1) with the arithmetic mean of the attribute. 

```python
import pandas as pd


df2 = pd.read_csv('Cereals1.csv')

columns_to_clean = ['protein', 'fat', 'sodium', 'fiber', 'carbo', 'sugars', 'potass', 'vitamins']

df2[columns_to_clean] = df2[columns_to_clean].apply(pd.to_numeric, errors='coerce')

df2[columns_to_clean] = df2[columns_to_clean].replace(-1, pd.NA)

for col in columns_to_clean:
    mean_val = df2[col].mean()
    df2[col] = df2[col].fillna(mean_val)


df2[columns_to_clean] = df2[columns_to_clean].round(2)


print(df2[columns_to_clean].head())    
print(df2.info())    
```

# 4) For each of the numeric attributes (proteins up to vitamins), identify and replace all 
noisy data with the median of attribute. 

```python
import pandas as pd
import numpy as np

# Load your dataset
df = pd.read_csv('Cereals.csv')

df.replace('?', np.nan, inplace=True) 

df = df.apply(pd.to_numeric, errors='coerce')

cols = df.columns.tolist()

start_idx = cols.index('protein')
end_idx = cols.index('vitamins') + 1
target_cols = cols[start_idx:end_idx]


for col in target_cols:
    median_val = df[col].median()
    df[col].fillna(median_val, inplace=True)


print(df[target_cols].describe().round(2))
```
